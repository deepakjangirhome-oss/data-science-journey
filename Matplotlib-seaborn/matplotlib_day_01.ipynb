```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Where we can use 2D plot
* Bivariate Analysis
* categorical -> numerical and numerical -> numerical
* Use case - Time series data

```python
# plotting a simple function
price = [35000, 41000, 48000, 44000, 41000, 30000]
year = [2015, 2016, 2017, 2018, 2019, 2020]

plt.plot(year, price)
plt.grid(True)
plt.xlabel('Year')
plt.ylabel('Price in $')
```

```python
# from a pandas dataframe
batsman = pd.read_csv("sharma-kohli.csv")
batsman

plt.plot(batsman['index'], batsman['V Kohli'])
plt.grid(True)
plt.xlabel('Year')
plt.ylabel('Run scored by Virat Kohli')
```

```python
# plotting multiple plots
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])

plt.grid(True)
plt.xlabel('Year')
plt.ylabel('Runs')
```

```python
# Lables title
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])

plt.title('Rohit Sharma vs Virat kohli carrier')
plt.xlabel('Season')
plt.ylabel('Runs Score')
```

```python
# color(hex) and line(width and style) and marker(size)
plt.plot(batsman['index'], batsman['V Kohli'], color='#4A0915')
plt.plot(batsman['index'], batsman['RG Sharma'], color='#1C6F85')

plt.title('Rohit Sharma vs Virat kohli carrier')
plt.xlabel('Season')
plt.ylabel('Runs Score')
```

```python
plt.plot(batsman['index'], batsman['V Kohli'], color='#4A0915', linestyle='solid', linewidth=3)
plt.plot(batsman['index'], batsman['RG Sharma'], color='#1C6F85', linestyle='dashdot', linewidth=2)

plt.title('Rohit Sharma vs Virat kohli carrier')
plt.xlabel('Season')
plt.ylabel('Runs Score')
```

```python
plt.plot(batsman['index'], batsman['V Kohli'], color='#4A0915', linestyle='solid', linewidth=3, marker='o', markersize=10)
plt.plot(batsman['index'], batsman['RG Sharma'], color='#1C6F85', linestyle='dashdot', linewidth=2, marker='o')

plt.title('Rohit Sharma vs Virat kohli carrier')
plt.xlabel('Season')
plt.ylabel('Runs Score')
```

```python
# legend -> location
plt.plot(batsman['index'], batsman['V Kohli'], color='#4A0915', linestyle='solid', linewidth=3, marker='o', markersize=10, label='Virat')
plt.plot(batsman['index'], batsman['RG Sharma'], color='#1C6F85', linestyle='dashdot', linewidth=2, marker='o', label = 'Rohit')

plt.title('Rohit Sharma vs Virat kohli carrier')
plt.xlabel('Season')
plt.ylabel('Runs Score')
plt.legend()
```

```python
# limiting axes
price = [35000, 41000, 48000, 44000, 41000, 30000, 4600000]
year = [2015, 2016, 2017, 2018, 2019, 2020, 2021]

plt.plot(year, price)
plt.ylim(0, 60000)
# plt.xlim(2017, 2020)
```

```python
# grid
# legend -> location
plt.plot(batsman['index'], batsman['V Kohli'], color='#4A0915', linestyle='solid', linewidth=3, marker='o', markersize=10, label='Virat')
plt.plot(batsman['index'], batsman['RG Sharma'], color='#1C6F85', linestyle='dashdot', linewidth=2, marker='o', label = 'Rohit')

plt.title('Rohit Sharma vs Virat kohli carrier')
plt.xlabel('Season')
plt.ylabel('Runs Score')
plt.legend()
plt.grid()
```

# Scatter plots

* Bivariate Analysis
* numerical vs numerical
* Use case - Finding correlation

```python
# plotting scatter plot
x = np.linspace(-10, 10, 50)

y = 10*x + 3 + np.random.randint(0,300,50)
y
```

```python
plt.scatter(x, y)
```

```python
# plt.scatter on pandas data
df = pd.read_csv('batter.csv')
df = df.head(50)
df
```

```python
# slower
plt.scatter(df['avg'], df['strike_rate'])
plt.title('Avg and SR analysis of top 50 batsman')
plt.xlabel('Average')
plt.ylabel('Strike Rate')
```

```python
# scatter plot using plt.plot
# faster
plt.plot(df['avg'], df['strike_rate'],'o')
```

# Bar Chart

* Bivariate Analysis
* Numerical vs Categorical
* Use case - Aggregate analysis of groups

```python
# Simple bar chart
children = [10, 20, 40, 25, 15]
colors = ['red', 'blue', 'green', 'yellow', 'pink']

plt.bar(colors, children)
```

```python
# Multiple bar charts
df = pd.read_csv('batsman_season_record.csv')
df
```

```python
plt.bar(np.arange(df.shape[0]) - 0.2, df['2015'],width=0.2)
plt.bar(np.arange(df.shape[0]), df['2016'],width=0.2)
plt.bar(np.arange(df.shape[0]) + 0.2, df['2017'],width=0.2)

plt.xticks(np.arange(df.shape[0]), df['batsman'])
plt.title('Runs analysis of seasons')
plt.xlabel('Players')
plt.ylabel('Runs')
```

```python
# a problem
children = [10, 20, 40, 25, 15]
colors = ['red red red red red red', 'blue blue blue blue blue', 'green green green green', 'yellow yellow yellow yellow', 'pink pink pink pink pink']

plt.bar(colors, children)
plt.xticks(rotation='vertical')
```

```python
# stacked bar chart
plt.bar(df['batsman'], df['2017'], label='2017')
plt.bar(df['batsman'], df['2016'],bottom = df['2017'], label='2016')
plt.bar(df['batsman'], df['2015'],bottom = (df['2016'] + df['2017']), label='2015')

plt.legend()
plt.show()
```

# Pie chart

* Univariate/Bivariate Analysis
* Categorical vs Numerical
* Use case - to fing contribution on a standard scale

```python
# simple chart
data = [23, 54, 34, 44, 17]
subjects = ['eng', 'math', 'hindi', 'sst', 'sanskrit']

plt.pie(data, labels=subjects)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# on a dataset
df = pd.read_csv('gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%')
plt.show()
```

```python
# percentage and colors
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%', colors=['blue', 'red', 'pink', 'cyan', 'yellow', 'black'])
plt.show()
```

```python
# explode and shadow
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%', explode=[0.1, 0,0,0,0,0], shadow=True)
plt.show()
```
